# Data exploration

Uses generated synthetic prices for software research only; these are not actual market data.

Inspect OHLCV rows and the data-quality report. Validation reports and normalizes; it does not silently repair or drop observations.


```python
from pathlib import Path
import sys
repo = Path.cwd()
while not (repo / "backend").is_dir() and repo != repo.parent:
    repo = repo.parent
sys.path.insert(0, str(repo / "backend"))

from trading.data.synthetic import generate_synthetic_ohlcv
from trading.data.validation import validate_ohlcv
bars = generate_synthetic_ohlcv(symbol="RELIANCE.NS", periods=500, frequency="1h", seed=7)
report = validate_ohlcv(bars, expected_interval="1h")
print({"rows": len(report.data), "symbols": report.data.symbol.unique().tolist(),
       "start": report.data.timestamp.min(), "end": report.data.timestamp.max(),
       "has_errors": report.has_errors,
       "issues": [(i.code, i.severity, i.row_count) for i in report.issues]})
bars.head()
```
